```python
#Sistema de gestão de notas

# Função para cadastrar as notas
def cadastrar_notas():
  notas = []
  quantidade = int(input("Quantas notas deseja cadastrar? ")) # Added space for better prompt reading

  for i in range(quantidade):
    # Loop until a valid float is entered
    while True:
      try:
        nota = float(input(f"Digite a {i + 1} nota: ")) # Added space
        notas.append(nota)
        break # Exit loop if input is valid
      except ValueError:
        print("Entrada inválida. Por favor, digite um número para a nota.")
  return notas # Moved return outside the loop

# Função para calcular a média
def calcular_media(notas):
  if not notas: # Handle case of empty list to avoid ZeroDivisionError
    return 0.0 # Or raise an error, depending on desired behavior
  media = sum(notas) / len(notas)
  return media

# Função para determinar a situação do aluno
def determinar_situacao(media):
  if media >= 7:
    return "Aprovado"
  else:
    return "Reprovado"

# Programa principal
notas = cadastrar_notas()
media = calcular_media(notas)
situacao = determinar_situacao(media) # Corrected function call

# Relatório principal
print("\n==== Relatório Final ====") # Corrected newline character
print("Notas inseridas:", notas)
print(f"Média: {media:.2f}")
print("Situação:", situacao)
```
